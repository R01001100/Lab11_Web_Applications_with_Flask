#### Try It: 

- Render the Form Fields in a Notebook

```python
# Try It: Render the Form Fields in a Notebook
from flask import Flask
from forms import RegisterForm

# Create temporary app
temp_app = Flask(__name__)
temp_app.config["SECRET_KEY"] = "temp-key"
temp_app.config["WTF_CSRF_ENABLED"] = False  # Disable CSRF for testing

# Need BOTH app context AND request context
with temp_app.app_context():
    with temp_app.test_request_context():  # This provides request context
        form = RegisterForm()
        print("Register Form Fields:")
        print(f"Email: {form.email}")
        print(f"Password: {form.password}")
        print(f"Submit: {form.submit}")

        # Verify your Data & Obs implementations
        print("\n--- Your Data & Obs Implementations Verified ---")
        print(
            f"✓ Email max length: {form.email.validators[2].max if len(form.email.validators) > 2 else 'N/A'} characters"
        )
        print(f"✓ Email description: {form.email.description}")
        print(
            f"✓ Password min length: {form.password.validators[1].min if len(form.password.validators) > 1 else 'N/A'}"
        )
```
